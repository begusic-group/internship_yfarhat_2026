```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
dt = 0.02
t = 1
nsteps = int(t/dt)
S = 6
```

```python
results = np.load('kagomem_repo32_results.npz')
string_counts = np.load('kagomem_repo32_string_counts.npz')
```

```python
times = np.arange(nsteps+1) * dt
plt.figure(figsize=(8, 5))


sorted_thresholds = sorted(results.keys(), key=float)
colors = plt.cm.rainbow(np.linspace(0, 1, len(sorted_thresholds)))

for i, thresh in enumerate(sorted_thresholds [:4]):
    thresh_val = float(thresh)
    exponent = int(np.log2(thresh_val))
    label = f'$2^{{{exponent}}}$'
    r = results[thresh]
    plt.plot(times, r, color = colors[i], linestyle = '-', label = label)

plt.ylabel(r'$\langle n_{8, \downarrow} \rangle$')
plt.xlabel(r' $\tau / t$')
plt.legend(loc='best')
plt.grid(True)
plt.tight_layout()
plt.show()

sorted_strings = sorted(string_counts.keys(), key=float)
colors = plt.cm.rainbow(np.linspace(0, 1, len(sorted_strings)))

for i, thresh in enumerate(sorted_thresholds [:4]):
    thresh_val = float(thresh)
    exponent = int(np.log2(thresh_val))
    label = f'$2^{{{exponent}}}$'
    r = results[thresh]
    plt.plot(times, r, color = colors[i], linestyle = '-', label = label)

plt.ylabel('# strings')
plt.xlabel(r'$\tau / t$')
plt.yscale('log')
plt.legend(loc='best')
plt.grid(True)
plt.tight_layout()
plt.show()
```
