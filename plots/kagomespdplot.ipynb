```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
dt = 0.12
t = 6
nsteps = int(t/dt)
threshold = 1e-5
```

```python
results = np.load('kagomespd.npz')
```

```python
times = np.arange(nsteps+1) * dt
plt.figure(figsize=(8, 5))


sorted_thresholds = sorted(results.keys(), key=float)
colors = plt.cm.rainbow(np.linspace(0, 1, len(sorted_thresholds)))

for i, thresh in enumerate(sorted_thresholds [:4]):
    thresh_val = float(thresh)
    exponent = int(np.log2(thresh_val))
    label = f'$2^{{{exponent}}}$'
    r = results[thresh]
    plt.plot(times, r, color = colors[i], linestyle = '-', label = label)

plt.ylabel(r'$\langle Z_{8} \rangle$')
plt.xlabel('Time')
plt.legend(loc='best')
plt.grid(True)
plt.tight_layout()
plt.show()
```
