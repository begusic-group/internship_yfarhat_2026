```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
dt = 0.12
t = 6
nsteps = int(t/dt)
threshold = 1e-5
```

```python
results = np.load('chainmajorana_finalnewresults.npz')
string_counts = np.load('chainmajorana_finalnewstring_counts.npz')
```

```python

times = np.arange(nsteps+1) * dt
plt.figure(figsize=(8, 5))

sorted_res = sorted(results.keys(), key=float)
colors = plt.cm.rainbow(np.linspace(0, 1, len(sorted_res)))

for i, key in enumerate(sorted_res):
    s_val = key.split('_')[0] 
    label = f'S = {s_val}'
    
    r = results[key]
    plt.plot(times, r, color=colors[i], linestyle='-', label=label)

plt.xlabel(r' $\tau / t$')
plt.ylabel(r'$\langle n_{50,\uparrow} \rangle$')
plt.legend(loc='best')
plt.grid(True)
plt.tight_layout()
plt.show()

sorted_strings = sorted(string_counts.keys(), key=float)
colors = plt.cm.rainbow(np.linspace(0, 1, len(sorted_strings)))

for i, key in enumerate(sorted_strings):
    s_val = key.split('_')[0]
    label = f'S = {s_val}'
    
    r = string_counts[key]
    plt.plot(times, r, color=colors[i], linestyle='-', label=label)

plt.ylabel('# strings')
plt.xlabel(r'$\tau / t$')
plt.yscale('log')
plt.legend(loc='best')
plt.grid(True)
plt.tight_layout()
plt.show()
```
