```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
dt = 0.12
t = 6
nsteps = int(t/dt)
threshold = 1e-5
```

```python
r = np.array(np.load('chainmajorana_fixed10results.npy'))
string_counts = np.load('chainmajorana_fixed10string_counts.npy')
```

```python
times = np.arange(nsteps + 1) * dt
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(times, r.real, 'o-', markersize=3)
plt.xlabel(r'Time $\tau / t$')
plt.ylabel(r'$\langle n_{50,\uparrow} \rangle$')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.subplot(1, 2, 2)
plt.semilogy(times, string_counts, 's-', markersize=3)
plt.xlabel(r'Time $\tau / t$')
plt.ylabel('# strings')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```
